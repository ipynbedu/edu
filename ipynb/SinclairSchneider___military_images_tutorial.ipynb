```
import numpy as np
from datasets import load_dataset, Image
from PIL import Image as PILImage
import random
import matplotlib.pyplot as plt
import pandas as pd

# ===========================================================================
# 📚 데이터셋 정보
# 데이터셋명: SinclairSchneider/military_images
# 주제: 군사 관련 이미지와 캡션 데이터셋
# 설명: 다양한 군사 작전, 장비, 상황을 담은 이미지와 이를 설명하는 텍스트가 쌍을 이루고 있습니다.
# 이 데이터셋으로 이미지 인식(CV)와 텍스트 분석(NLP)을 결합하는 멀티모달 실습을 해봅시다! 🚀
# ----------------------------------------------------------------------------

# 사용할 데이터셋 ID와 스플릿을 정의합니다.
DATASET_NAME = "SinclairSchneider/military_images"
SPLIT_NAME = "train" # 학습용 데이터셋을 사용합니다.
SAMPLE_COUNT = 10 # 초보자 실습이므로, 처음 10개의 샘플만 가져와서 빠르게 테스트합니다!

print("🤖✨ 안녕하세요! 저는 여러분의 친절하고 위트 있는 AI 코딩 튜터입니다! ✨🤖")
print(f"오늘은 '군사 이미지(military_images)' 데이터셋을 가지고 멀티모달 AI 실습을 해볼 거예요. 쉽고 재미있게 따라와 주세요!")
print("-" * 70)

# 1. 데이터 로드 전략: 스트리밍을 우선시합니다!
try:
    print(f"✅ [1/3] 데이터셋 로드 시도: '{DATASET_NAME}', {SPLIT_NAME} (스트리밍 모드)")
    # streaming=True를 사용하여 메모리 부담 없이 스트리밍으로 데이터를 가져옵니다.
    dataset = load_dataset(DATASET_NAME, split=SPLIT_NAME, streaming=True)
    print("🎉 성공! 스트리밍 모드로 데이터셋을 로드했습니다. 메모리 효율 최고!")

except Exception as e:
    print(f"⚠️ 스트리밍 로드 실패 또는 환경 제약 발생: {e}")
    print("🔄 일반 모드(Non-streaming)로 fallback하여 로드합니다. (메모리 주의!)")
    try:
        # 스트리밍이 안되면 일반 모드로 로드합니다.
        dataset = load_dataset(DATASET_NAME, split=SPLIT_NAME)
        print("✅ 일반 모드로 데이터셋을 로드했습니다. 이제 실습할 준비가 끝났어요!")
    except Exception as e_fail:
        print(f"💥 치명적인 오류가 발생하여 데이터셋 로드에 실패했습니다: {e_fail}")
        print("👉 프로그램 종료합니다.")
        raise SystemExit("데이터 로드 실패!")


# 2. 샘플 데이터 추출 준비 (핵심 패턴 준수)
# 데이터셋이 스트리밍(IterableDataset)인지 일반(Dataset)인지 확인하고 샘플을 가져올 준비를 합니다.
print("\n--- 샘플 데이터 추출 준비 ---")

if hasattr(dataset, "take"):
    # .take()가 존재하면 스트리밍 데이터셋 (IterableDataset)입니다.
    print("➡️ 스트리밍 데이터셋 감지: .take()를 사용하여 상위 샘플을 가져옵니다.")
    # take()를 사용하여 상위 K개의 샘플만 가져옵니다.
    sample_iterator = dataset.take(SAMPLE_COUNT)
    # list(iterator) 패턴을 사용하여 모든 샘플을 메모리에 로드합니다.
    sample_data_list = list(sample_iterator)
else:
    # 일반 데이터셋 (Dataset)입니다.
    print("➡️ 일반 데이터셋 감지: 리스트 변환을 통해 상위 샘플을 가져옵니다.")
    sample_data_list = list(dataset.select(range(SAMPLE_COUNT)))

if not sample_data_list:
    print("❌ 가져온 샘플 데이터가 없습니다. 스크립트를 종료합니다.")
    exit()

print(f"✨ 최종적으로 {len(sample_data_list)}개의 샘플을 가지고 실습을 진행할게요!")
sample_dataset = sample_data_list

# 3. 메인 실습 함수
def run_multimodal_analysis(sample_data):
    """
    이미지 데이터와 텍스트 데이터를 활용하여 창의적인 분석을 수행합니다.
    """
    print("\n" + "=" * 70)
    print("🧠 [2/3] 실습 1: 데이터 구조 이해 및 이미지 전처리 (Vision Task)")
    print("=" * 70)

    # 💡 목표: 이미지 데이터의 크기를 확인하고 NumPy 배열로 변환하는 법을 배웁니다.
    print("✨ [튜터 Tip] 허깅페이스에서 로드된 이미지는 PIL Image 객체인 경우가 많아요. "
          "NumPy 배열로 다룰 수 있게 변환해 봅시다!")

    image_samples = []
    for i, sample in enumerate(sample_data):
        image = sample["image"]
        # PIL Image 객체 -> NumPy 배열 변환 (CV 필수 과정)
        # .convert('RGB')를 사용해 채널 정보를 명확히 합니다.
        np_image = np.array(image)
        image_samples.append(np_image)

    # 첫 번째 샘플의 이미지 크기 확인
    if image_samples:
        sample_img_np = image_samples[0]
        print(f"\n🖼️ 첫 번째 이미지의 NumPy 형태 (Shape): {sample_img_np.shape} (Height, Width, Channels)")

        # 🔬 창의적 실습 1-1: 이미지 로드된 데이터 분석 및 시각화
        plt.figure(figsize=(10, 5))
        # 이미지 배열을 Matplotlib로 표시합니다.
        plt.subplot(1, 2, 1)
        plt.imshow(sample_img_np)
        plt.title("Sample Image (Visual Check)")
        plt.axis('off')
        plt.subplot(1, 2, 2)
        # 첫 번째 텍스트를 가공하여 표시합니다.
        text_sample = sample_data[0]["text"]
        print(f"📝 첫 번째 캡션 길이: {len(text_sample)} 문자")
        plt.text(0.05, 0.8, f"Text Preview:\n{text_sample[:100]}...", 
                 transform=plt.gca().transAxes, ha='left', va='top', fontsize=10)
        plt.title("Textual Context")
        plt.axis('off')
        plt.suptitle(f"Sample {0} (Image: {sample_img_np.shape[0]}x{sample_img_np.shape[1]}, Text Length: {len(text_sample)})", fontsize=14)
        plt.show()
    else:
        print("이미지 데이터를 추출할 수 없습니다.")


    print("\n" + "=" * 70)
    print("✨ [3/3] 실습 2: 캡션(Text)을 활용한 키워드 기반 '간단 분류 시뮬레이션' (NLP Task)")
    print("=" * 70)
    
    # 💡 목표: 이미지와 텍스트를 모두 활용하여 AI가 '어떤 작전'에 대한 내용인지 추측하는 시뮬레이션입니다.
    
    keyword_scores = []
    keywords = ["ground", "air", "sea", "equipment"]
    
    print("🔍 키워드를 분석하여 각 샘플이 어떤 유형의 작전을 묘사하는지 점수를 매겨봅시다.")

    for sample in sample_data:
        text = sample["text"].lower()
        score = {}
        total_score = 0
        
        for keyword in keywords:
            # 키워드가 포함되어 있는지 확인하고 점수를 부여합니다.
            if keyword in text:
                score[keyword] = text.count(keyword)
                total_score += 1
            else:
                score[keyword] = 0
        
        # 텍스트를 분석한 결과를 저장합니다.
        keyword_scores.append(score)
        
        # 간단한 결과 출력 (과부하 방지 위해 처음 5개만 보여줍니다)
        if len(keyword_scores) < 5:
            main_keyword = max(score, key=score.get)
            print(f"   -> 📝 '{sample['text'][:40]}...' | 가장 높은 점수 키워드: {main_keyword.upper()} (Score: {score[main_keyword]})")

    print("\n✅ 분석 완료! 키워드 기반으로 작전의 성격을 파악할 수 있었어요. AI가 이렇게 작동합니다!")
    
    
def main():
    """메인 실행 함수"""
    run_multimodal_analysis(sample_dataset)
    print("\n\n🎉🎉🎉 축하합니다! 🎉🎉🎉")
    print("멀티모달 데이터셋을 로드하고, 시각화하며, 텍스트 분석까지 성공적으로 해냈어요!")
    print("이것이 바로 데이터 과학의 재미입니다! 계속 코딩하세요!")

if __name__ == "__main__":
    # matplotlib 사용을 위해 기본 설정합니다.
    plt.ion() # 인터랙티브 모드 활성화 (출력물이 잘 보이게)
    main()
    plt.ioff() # 작업 완료 후 비활성화
```

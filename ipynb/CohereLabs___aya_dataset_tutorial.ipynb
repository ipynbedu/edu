```
import random
import pandas as pd
from datasets import load_dataset, get_dataset_config_names

# ==========================================================================
# 🧑‍🏫 AI 튜터의 코딩 강의 시작! 💡
# 📚 데이터셋 제목: Aya Dataset (아야 데이터셋)
# 📚 의미: 전 세계 수많은 언어(20개 이상!)의 다양한 문장 쌍(입력/타겟)으로 구성된 초거대 다국어 말뭉치입니다.
# 📚 설명: 마치 지구상의 모든 언어가 모여있는 거대한 도서관 같아요. 우리는 이 데이터셋을 사용해 '어떻게 다국어 데이터를 효율적으로 탐색하고 구조화할까?'를 배워볼 거예요!
# ==========================================================================

# 목표: 초보자가 다국어 데이터셋의 구조를 탐색하는 창의적 실습
DATASET_NAME = "CohereLabs/aya_dataset"
TEST_SPLIT = 'test'
SAMPLE_COUNT = 10 # 분석을 위해 상위 10개 샘플만 사용합니다! (시간 절약 꿀팁!)

# 1. 데이터셋의 사용 가능한 설정 확인 (필수 단계!)
print("=" * 80)
print(f"✨ [Step 0] {DATASET_NAME}의 사용 가능한 설정을 확인해 봅시다!")
try:
    configs = get_dataset_config_names(DATASET_NAME)
    print(f"✅ 사용 가능한 Config 목록: {configs}")
    selected_config = configs[0] if configs else None
except Exception as e:
    print(f"ℹ️ Config 조회 오류 발생: {e}. 기본 설정을 사용합니다.")
    selected_config = None

# 2. 데이터 로딩 전략 (스트리밍 우선!)
dataset = None
print("\n" + "=" * 80)
print("🚀 [Step 1] 데이터셋 로딩 시작: 메모리 효율을 위해 '스트리밍' 모드를 먼저 시도합니다!")

try:
    # Constraint 2: streaming=True로 로딩 시도
    dataset = load_dataset(DATASET_NAME, split=TEST_SPLIT, streaming=True)
    print("🎉 성공! 스트리밍 모드(Streaming=True)로 데이터를 로드했습니다. 메모리를 아끼는 똑똑한 방법이에요!")
except Exception as e:
    print(f"⚠️ 스트리밍 로드 실패 ({type(e).__name__}). 에러 발생! ({e})")
    print("🔄 [대안] 스트리밍이 어렵다면, 적은 양만 다운로드하여 진행합니다.")
    # Constraint 3: 스트리밍 실패 시, 작은 양만 다운로드
    try:
        dataset = load_dataset(DATASET_NAME, split=TEST_SPLIT, streaming=False)
        print("💾 일반 다운로드 모드(Streaming=False)로 테스트 스플릿을 로드했습니다. 이제 분석할 준비 완료!")
    except Exception as e_fallback:
        print(f"🚨 치명적 에러: 데이터셋 로드에 실패했습니다. {e_fallback}")
        exit()

# 3. 데이터 샘플링을 위한 iterator 설정 (가장 까다로운 부분!)
print("\n" + "=" * 80)
print("✨ [Step 2] 데이터 샘플링 및 이터레이터 설정 (K=10)")

# Constraint 9 패턴 적용: take() 메소드 확인
if hasattr(dataset, "take"):
    # .take()가 존재하면 스트리밍 데이터셋(IterableDataset)
    print("🧠 스트리밍 데이터셋(IterableDataset) 패턴 감지. take() 메소드를 사용합니다.")
    # Constraint 9 패턴을 따릅니다.
    sampled_dataset_iterator = iter(dataset.take(SAMPLE_COUNT))
else:
    # 일반 데이터셋 (Dataset)
    print("📚 일반 Dataset 패턴 감지. .take()를 사용합니다.")
    # Constraint 9 패턴을 따릅니다.
    sampled_dataset = dataset.take(SAMPLE_COUNT)
    sampled_dataset_iterator = iter(sampled_dataset)


# 4. 데이터 샘플 분석 및 창의적 실습 (데이터 프로파일링 시뮬레이션)
print("\n" + "=" * 80)
print("🌟 [Step 3] 🎉 실습: 다국어 데이터 구조 분석 (데이터 프로파일링)")
print("===============================================================================================================")

# 샘플 데이터를 가져와 리스트로 저장합니다. (Constraint 16 적용)
sample_data_list = list(sampled_dataset_iterator)

if not sample_data_list:
    print("🚫 분석할 샘플 데이터가 없습니다. 데이터 로딩을 확인해 주세요.")
else:
    print(f"✅ 성공! 상위 {len(sample_data_list)}개의 샘플 데이터를 메모리에 로드했습니다.")

    # 4-1. 데이터 구조 파악 (첫 번째 샘플 확인)
    sample_1 = sample_data_list[0]
    print("\n--- 🔍 [탐색 1] 데이터의 기본 구조 확인 (가장 첫 번째 샘플) ---")
    print(f"  - 전체 샘플 타입: {type(sample_1)}")
    print(f"  - 보유 Feature: language, language_code, inputs, targets, annotation_type, user_id")
    print(f"  - 예시 언어: {sample_1['language']} (코드: {sample_1['language_code']})")
    print(f"  - 인풋 문장: '{sample_1['inputs'][:30]}...'")


    # 4-2. 다국어 특성 분석 (Language & Language Code 매핑)
    print("\n--- 🌍 [탐색 2] 언어 다양성 분석 (Language vs Code) ---")
    language_counts = {}
    for sample in sample_data_list:
        lang = sample['language']
        lang_code = sample['language_code']
        # 딕셔너리에 언어별 등장 횟수를 기록합니다.
        language_counts[lang] = language_counts.get(lang, 0) + 1
    
    print(f"  ✅ 상위 {len(sample_data_list)}개 샘플에서 감지된 고유 언어 수: {len(language_counts)}개")
    # 가장 많이 나온 언어 3개를 뽑아 출력해봅니다.
    sorted_languages = sorted(language_counts.items(), key=lambda item: item[1], reverse=True)
    print("  ➡️ 가장 많이 등장한 언어 3가지 (샘플 기준):")
    for lang, count in sorted_languages[:3]:
        print(f"    - {lang}: {count}회 등장")


    # 4-3. 데이터의 사용 목적 추론 (LLM Prompt 생성 시뮬레이션)
    print("\n--- 🛠️ [창의 실습] AI 모델링을 위한 데이터 준비 시뮬레이션 ---")
    print("  💡 목표: 우리는 이 데이터 쌍(inputs -> targets)을 이용하여 특정 언어에 맞는 '프롬프트 템플릿'을 만들고 싶습니다.")

    # 특정 언어(예: 영어 'eng') 필터링 및 구조화
    english_samples = [s for s in sample_data_list if s['language'] == 'eng']
    
    if english_samples:
        print(f"  🔍 'eng' (영어) 언어 데이터 ({len(english_samples)}개) 필터링 완료.")
        print("  - 데이터 분석 결과: 이 데이터셋은 단순한 번역 쌍이 아니라, 다양한 맥락의 문장(inputs)이 같은 의미의 문장(targets)으로 변환되는 패턴을 포함하고 있어요!")

        # 데이터 구조화를 위한 간단한 루프 (list comprehension 활용)
        processed_examples = []
        for sample in english_samples[:3]: # 상위 3개만 사용
            # 튜플 형태로 (입력문장, 타겟문장, 언어)를 추출
            processed_examples.append((sample['inputs'], sample['targets'], sample['language']))
        
        print("\n  ✨ 준비된 튜플 리스트 (모델 학습에 사용할 핵심 데이터):")
        for i, example in enumerate(processed_examples):
            print(f"    [{i+1}] (Input: {example[0][:40]}...) -> (Target: {example[1][:40]}...) | Lang: {example[2]}")
    else:
        print("  🤔 영어('eng') 데이터를 가진 샘플을 찾지 못했습니다. 데이터를 더 탐색해 보세요!")

print("\n" + "=" * 80)
print("💖 축하합니다! 당신은 다국어 데이터셋의 구조를 성공적으로 분석했습니다.")
print("    데이터를 로드하는 것만큼, '어떤 정보를 추출해야 할지' 아는 것이 AI 엔지니어링의 핵심이랍니다! 👍")
```

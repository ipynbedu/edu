## 1. 변수(Variable)란?

변수는 **값에 이름을 붙여서 저장해두는 것**입니다. 상자에 이름표를 붙이는 것과 같아요.

```python
age = 25
```

이 코드는 "`25`라는 값을 `age`라는 이름의 상자에 넣어라"라는 뜻입니다. 이후에 `age`라고 쓰면 그 상자 안의 값(`25`)을 그대로 꺼내 쓸 수 있습니다.

직접 실행해서 확인해봅시다.

````python
age = 25
print(age)
````

`print()`는 괄호 안에 든 값을 화면에 보여달라는 명령입니다. 변수를 만들기만 하고 `print()`를 안 하면, 값이 상자 안에 들어있긴 해도 화면엔 아무것도 안 보입니다.

**주의할 점 하나.** `=`는 수학에서 배운 "같다"가 아닙니다. 파이썬에서 `=`는 **"오른쪽 값을 왼쪽 이름에 넣어라"**는 뜻입니다. ("같다"를 뜻하는 기호는 나중에 U3에서 따로 배웁니다 — 지금은 헷갈리지 않게 `=`은 무조건 "넣어라"로만 기억하세요.)

## 2. 자료형(Data Type) — 값에는 "종류"가 있다

| 자료형 | 이름 | 예시 |
|---|---|---|
| `int` | 정수 (소수점 없는 숫자) | `25`, `-3`, `0` |
| `float` | 실수 (소수점 있는 숫자) | `3.14`, `-0.5` |
| `str` | 문자열 (글자) — 반드시 따옴표로 감쌈 | `"안녕"`, `'25'` |
| `bool` | 불리언 (참/거짓, 둘 중 하나만) | `True`, `False` |

**중요한 함정 하나.** `25`(숫자)와 `"25"`(문자열)는 겉보기엔 비슷해 보여도 완전히 다른 것입니다. `"25"`는 계산할 수 없는 그냥 글자 두 개일 뿐입니다.

### ① 개념 · 예측하기

아래 코드를 **실행하기 전에**, `type(...)`의 결과가 각각 무엇일지 먼저 종이에 적어보세요. (`type()`은 "이 값이 무슨 자료형이야?"라고 물어보는 명령입니다. 결과는 `<class '자료형이름'>` 모양으로 나옵니다.)

````python
age = 25          # 정수
height = 172.5    # 실수
name = "민수"      # 문자열
is_adult = True   # 불리언

print(type(age))
print(type(height))
print(type(name))
print(type(is_adult))
````

예측을 다 적었다면 위 셀을 직접 실행해보세요. 실행 결과가 아래와 같이 나오면 정상입니다.

```
<class 'int'>
<class 'float'>
<class 'str'>
<class 'bool'>
```

예측이 맞았나요? 맞았다면 자료형 구분 감각이 이미 잡힌 것입니다.

```
import random
from datasets import load_dataset, Dataset, IterableDataset
import os

# ==============================================================================
# 🏆 AI 튜터의 코딩 팁: 데이터셋 분석 및 준비
# 🌟 데이터셋명: yang-ki/army_sample
# 🧐 데이터 의미: 이 데이터셋은 마치 '질문-입력-답변'의 튜토리얼 자료처럼 구성되어 있습니다.
#    'instruction' (지시문)과 'input' (추가 정보)을 주고, 'output' (최종 답)을 유도하는
#    프롬프트 엔지니어링 훈련용 데이터셋입니다.
# 💡 우리의 미션: 초급자가 이 데이터를 가지고 '가장 최적의 질문 프롬프트'를 만드는 방법을 연습해봅시다!
# ==============================================================================

# 🚀 글로벌 설정 및 상수 정의
DATASET_NAME = "yang-ki/army_sample"
# 🔥 너무 많은 데이터를 로드하면 시간이 오래 걸리니, 일단 10개 샘플로 테스트합니다!
SAMPLE_COUNT = 10 

print("=====================================================")
print("✨ 환영합니다! 친절하고 위트 넘치는 AI 코딩 튜터가 당신의 코딩 여정을 도와줄게요!")
print("=====================================================")

# 1. 데이터 로드 및 스트리밍 테스트 (⚡️ 빠르고 효율적인 로딩이 핵심!)
dataset = None
sample_data_list = []
loaded_type = "N/A"

print(f"\n[STEP 1] '{DATASET_NAME}' 데이터셋을 로드합니다. 스트리밍 모드를 먼저 시도할게요!")

# 🟢 필수 요구사항 2, 3: 스트리밍 로드 시도 (가장 빠름!)
try:
    # split='train' 명시, streaming=True로 설정
    dataset = load_dataset(DATASET_NAME, split='train', streaming=True)
    loaded_type = "Streaming (IterableDataset)"
    print("✅ 성공! 스트리밍 모드로 데이터셋을 로드했습니다. 메모리 효율 최고!")
except Exception as e:
    print(f"🚨 스트리밍 로드에 실패했습니다. 오류: {e}")
    print("➡️ 대안으로, 작은 수의 데이터만 다운로드하여 일반 Dataset 모드로 로드할게요.")
    try:
        # 🟡 실패 시 대안: 소량의 데이터를 일반 Dataset으로 로드
        dataset = load_dataset(DATASET_NAME, split='train')
        loaded_type = "Standard (Dataset)"
    except Exception as e_fallback:
        print(f"❌ 모든 로드 시도가 실패했습니다. 코드를 종료합니다. 오류: {e_fallback}")
        exit()

# 2. 샘플 데이터 준비 (Memory Safety & Streaming Guard)
print("\n[STEP 2] 학습에 사용할 샘플 데이터만 추출합니다.")

# 🟢 필수 요구사항 9: 스트리밍/일반 데이터셋 모두 적용되는 안전한 패턴 사용
if hasattr(dataset, "take"):
    # .take()가 존재하면 스트리밍 데이터셋(IterableDataset)입니다.
    print(f"📚 {loaded_type} 모드 감지: take() 메소드를 사용합니다.")
    
    # list() 함수를 사용하여 안전하게 샘플 N개를 메모리에 올립니다.
    sample_data_list = list(dataset.take(SAMPLE_COUNT))
else:
    # 일반 Dataset인 경우
    sample_data_list = dataset.select(range(min(SAMPLE_COUNT, len(dataset))))

# 📊 정량적 분석: 우리가 사용할 데이터의 양을 확인해봅시다!
print(f"💡 준비 완료! 총 {len(sample_data_list)}개의 샘플을 사용하여 실습을 진행할 거예요.")


# 3. 실습 시작: 프롬프트 구조화 및 'AI 예측' 시뮬레이션
print("\n" + "="*70)
print(f"🚀 [STEP 3] AI 프롬프트 구조화 마스터 실습 시작! ({len(sample_data_list)}개 샘플)")
print("=====================================================")

processed_count = 0

print("✨ 튜터의 가이드: 데이터셋은 '지시(Instruction)'와 '입력(Input)'을 통해 '출력(Output)'을 만들어내는 과정을 담고 있어요.")
print("   우리는 마치 LLM (대규모 언어 모델)의 '선생님'이 되어, 최적의 프롬프트를 짜주는 연습을 해볼 거예요!")
print("-" * 70)

# 데이터셋 전체를 루프 돌리는 대신, 준비된 샘플 리스트만 사용합니다.
for i, sample in enumerate(sample_data_list):
    processed_count += 1
    
    instruction = sample['instruction']
    input_text = sample['input']
    output = sample['output']
    
    # 🧠 실습 목표: I, I, O 세 요소를 조합하여 '최종 프롬프트'를 만드는 방법을 배웁니다.
    # 이 과정 자체가 AI 엔지니어의 핵심 역량 중 하나입니다!
    
    print(f"\n{'='*5} [샘플 #{i+1} / {len(sample_data_list)}] {'='*5}")
    
    # 📝 구조화된 프롬프트 만들기 (Prompt Engineering)
    # input이 비어있지 않을 때만 input을 프롬프트에 포함하도록 조건문을 씁니다.
    if input_text:
        full_prompt = f"[INSTRUCTION] {instruction}\n\n[INPUT DATA] {input_text}\n\n[TASK]: "
    else:
        full_prompt = f"[INSTRUCTION] {instruction}\n\n[TASK]: "
        
    print(f"✍️ ✨ 최종 프롬프트 생성 ✨ (LLM에 던져줄 질문):")
    print(f"----------------------------------------------------")
    print(full_prompt.strip())
    print("----------------------------------------------------")

    # 🔮 AI 모델처럼 '예측'하기 (실제로는 LLM API 호출이 필요하지만, 여기선 데이터셋의 'output'을 사용합니다.)
    # 이는 "만약 위 프롬프트를 던졌다면, 이렇게 답할 것이다"라고 예상하는 과정입니다.
    print("\n🤖 LLM 시뮬레이션 결과 (가장 기대되는 정답):")
    print(f">>> {output.strip()}")
    
    # 잠시 멈춤 효과를 주어 학습자의 집중도를 높입니다.
    # time.sleep(0.1) 

print("\n" + "="*70)
print("🎉 실습 완료! 축하합니다!")
print(f"✨ 총 {processed_count}개의 샘플을 분석하여 완벽한 프롬프트 구조화 능력을 익혔습니다.")
print("💪 데이터셋의 각 요소를 이해하고, 주어진 목적에 맞게 조합하는 능력이야말로 AI 활용의 기본 중 기본입니다!")
print("궁금한 점이 있다면 언제든지 저에게 질문해주세요! 😊")
print("="*70)
```

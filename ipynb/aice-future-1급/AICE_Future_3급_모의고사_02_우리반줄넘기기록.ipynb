## 채점

```python
# 채점: 문제를 푼 뒤(또는 해설 셀을 실행한 뒤) 이 셀을 실행하세요.
_key = {
    1: 1,
    2: 2,
    3: 3,
    4: 1,
    5: 1,
    6: ['A', 'B', 'C', 'D'],
    7: 1,
    8: 1,
}
_points = {
    1: 10,
    2: 15,
    3: 15,
    4: 15,
    5: 10,
    6: 15,
    7: 10,
    8: 10,
}
_pass_score = 60
_total = 100
_score = sum(_points[n] for n, correct in _key.items() if answers.get(n) == correct)
_result = '합격' if _score >= _pass_score else '불합격'
print(f'3급 모의고사 채점 결과: {_score} / {_total}점 -> {_result} (합격선 {_pass_score}점)')
for n in sorted(_key):
    ok = '✅' if answers.get(n) == _key[n] else ('⬜' if answers.get(n) is None else '❌')
    print(f'  {ok} {n}번  내 답: {answers.get(n)}  /  정답: {_key[n]}')
```

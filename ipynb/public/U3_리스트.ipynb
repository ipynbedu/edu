## 리스트란?

리스트는 **여러 값을 하나로 묶어 담는 것**입니다. 대괄호 `[ ]`로 감싸고, 값 사이는 쉼표로 구분합니다.

실제 문항 10-1은 DecisionTree의 `max_depth`(트리 최대 깊이) 후보를 이렇게 씁니다: `[3, 5, 7, 10]`.

```python
max_depth_candidates = [3, 5, 7, 10]
print(max_depth_candidates[0])
```

대괄호 안의 숫자는 **인덱스(위치 번호)**이고, 파이썬은 **0부터** 셉니다. 그래서 `[0]`은 첫 번째 값 `3`입니다.

### ① 예측하기

실행 전에 결과를 먼저 예상해보세요.

```python
max_depth_candidates = [3, 5, 7, 10]
print(max_depth_candidates[0])
print(max_depth_candidates[2])
```

정답: `3`, `7`. **인덱스 2의 값**을 이 문제의 최종 답으로 저장합니다.

```python
정답01 = max_depth_candidates[2]
print(정답01)
```

### ② 빈칸 채우기

`10`이 출력되도록 `______` 자리에 들어갈 인덱스(숫자)를 채우세요.

```python
max_depth_candidates = [3, 5, 7, 10]
print(max_depth_candidates[______])
```

```python
정답02 = 3
print(정답02)
```

값이 4개(`3,5,7,10`)라 인덱스는 `0,1,2,3`까지만 있습니다. 마지막 값(`10`)의 인덱스는 `3`입니다.

### ③ 오류 고치기

아래 코드는 에러가 납니다. 왜인지 생각해보고 고쳐보세요.

```python
max_depth_candidates = [3, 5, 7, 10]
print(max_depth_candidates[4])
```

값이 4개면 인덱스는 `0~3`까지만 존재합니다. `[4]`는 존재하지 않는 다섯 번째 자리를 요구한 것입니다.

```python
max_depth_candidates = [3, 5, 7, 10]
정답03 = max_depth_candidates[3]
print(정답03)
```

## 변수란?

변수는 **값에 이름을 붙여서 저장하는 것**입니다 — 상자에 이름표를 붙이는 것과 같습니다. 데이터의 두 번째 행(와인 하나)에는 이런 값이 있습니다: `E_country`(생산국가) `Portugal`, `E_price`(가격) `15.0`, `E_FE_vintage`(수확연도) `2011`, `Grade`(품질등급) `Fail`.

```python
E_country = "Portugal"
print(E_country)
```

`print()`는 값을 화면에 보여주는 명령이고, `=`는 "같다"가 아니라 **"오른쪽 값을 왼쪽 이름에 넣어라"**는 뜻입니다.

## 자료형 4가지

| 자료형 | 뜻 | 예 |
|---|---|---|
| `int` | 정수 | `E_FE_vintage` → `2011` |
| `float` | 실수(소수점 O) | `E_price` → `15.0` |
| `str` | 문자열, 따옴표 필수 | `E_country` → `"Portugal"` |
| `bool` | 참/거짓 | 등급이 Fail인지 → `True`/`False` |

`2011`(숫자)과 `"2011"`(문자열)은 다른 것입니다 — 후자는 계산 불가능한 글자입니다.

### ① 예측하기

실행 전에 `type(...)` 결과 4개를 먼저 예상해보세요.

```python
E_country = "Portugal"
E_price = 15.0
E_FE_vintage = 2011
is_fail = True

print(type(E_country))
print(type(E_price))
print(type(E_FE_vintage))
print(type(is_fail))
```

정답: `str`, `float`, `int`, `bool`. 이 중 **int형은 몇 개**인지가 이 문제의 최종 답입니다.

```python
정답01 = 1
print(정답01)
```

### ② 빈칸 채우기

`Portugal`이 출력되도록 `______`를 채우세요.

```python
정답02 = "="
print(정답02)
```

### ③ 오류 고치기

아래 코드는 에러가 납니다. 왜인지 생각해보고 고쳐보세요.

```python
E_country = Portugal
print(E_country)
```

따옴표 없는 `Portugal`은 존재하지 않는 변수 이름으로 취급됩니다. 문자열엔 따옴표가 필수입니다.

```python
E_country = "Portugal"
정답03 = E_country
print(정답03)
```

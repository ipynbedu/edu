## float란?

`float`은 소수점이 있는 숫자입니다. 데이터의 두 번째 행(와인 하나)의 가격(`E_price`)은 `15.0`입니다.

```python
E_price = 15.0
print(E_price)
```

### ① 예측하기

U1에서 배운 것까지 합쳐서 3개입니다. 실행 전에 `type(...)` 결과를 먼저 예상해보세요.

```python
E_country = "Portugal"
E_FE_vintage = 2011
E_price = 15.0

print(type(E_country))
print(type(E_FE_vintage))
print(type(E_price))
```

정답: `str`, `int`, `float`. 이 중 **float형은 몇 개**인지가 이 문제의 최종 답입니다.

```python
정답01 = 1
print(정답01)
```

### ② 빈칸 채우기

`15.0`이 출력되도록 `______`를 채우세요.

```python
E_price ______ 15.0
print(E_price)
```

```python
정답02 = "="
print(정답02)
```

### ③ 오류 고치기

아래 코드는 에러가 납니다. 왜인지 생각해보고 고쳐보세요.

```python
E_price = "15.0"
print(E_price + 1)
```

따옴표를 씌운 `"15.0"`은 숫자가 아니라 문자열입니다. 글자와 숫자는 더할 수 없습니다 — 따옴표를 빼서 진짜 `float`로 만들어야 계산이 됩니다.

```python
E_price = 15.0
정답03 = E_price + 1
print(정답03)
```

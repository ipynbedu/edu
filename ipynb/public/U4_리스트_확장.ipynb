# 🔰 U4. 리스트 확장 — len()

U3에서 만든 리스트에 값이 몇 개 들어있는지 세는 함수, **`len()`** 하나만 추가합니다. 값은 이번에도
실제 AICE Associate 공식 샘플문항에서 그대로 가져왔습니다.

- 최종 답은 `정답01`, `정답02`... 순서로 담습니다.
- 다 풀었으면 노트북을 닫고 맨 아래 문제를 안 보고 다시 풀어보세요.

## len()이란?

리스트에 값이 몇 개 있는지 세어주는 명령입니다. 문항 10-1의 `max_depth` 후보는 `[3, 5, 7, 10]`, 총 4개입니다.

```python
max_depth_candidates = [3, 5, 7, 10]
print(len(max_depth_candidates))
```

### ① 예측하기

실행 전에 결과를 먼저 예상해보세요.

```python
n_estimators_candidates = [100, 200, 500]
print(len(n_estimators_candidates))
print(n_estimators_candidates[0])
```

정답: `3`, `100`. **`len()` 결과**를 이 문제의 최종 답으로 저장합니다.

```python
정답01 = len(n_estimators_candidates)
print(정답01)
```

### ② 빈칸 채우기

리스트의 값 개수가 출력되도록 `______`를 채우세요.

```python
max_depth_candidates = [3, 5, 7, 10]
print(______(max_depth_candidates))
```

```python
정답02 = "len"
print(정답02)
```

### ③ 오류 고치기

아래 코드는 에러가 납니다. 왜인지 생각해보고 고쳐보세요.

```python
max_depth_candidates = [3, 5, 7, 10]
print(len(max_depth_candidates()))
```

괄호 `()`는 리스트가 아니라 함수 뒤에 붙이는 것입니다. `max_depth_candidates()`처럼 리스트 이름 뒤에 괄호를 붙이면 "리스트를 함수처럼 호출하라"는 뜻이 되어 에러가 납니다. 괄호는 `len()` 쪽에만 있어야 합니다.

```python
max_depth_candidates = [3, 5, 7, 10]
정답03 = len(max_depth_candidates)
print(정답03)
```

### ④ 직접 작성하기

실제 문항 7-3은 삭제할 컬럼 5개를 이렇게 리스트로 씁니다: `['E_province', 'E_region_1', 'E_region_2', 'E_winery', 'E_wine_variety']`. 이 리스트를 만들고 `len()`으로 개수를 구하는 코드를 작성하세요. 그 값을 최종 답으로 저장하세요.

```python
# 여기에 코드를 작성하세요
```

```python
drop_columns = ['E_province', 'E_region_1', 'E_region_2', 'E_winery', 'E_wine_variety']
정답04 = len(drop_columns)
print(정답04)
```

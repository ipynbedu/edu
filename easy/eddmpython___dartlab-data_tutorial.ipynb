```
import re
from datasets import load_dataset
from typing import Iterator

# ==============================================================================
# 🌟 DartLab 공시 데이터 탐험가 되기 🌟
# 데이터셋: eddmpython/dartlab-data
# 주제: 기업 공시 문서에서 핵심 키워드(예: '매출', '수익')를 찾아 빈도를 분석해봅시다.
# 목표: 파이썬 스트리밍 데이터셋을 다루는 기본 원리와 간단한 텍스트 분석을 경험합니다!
# ==============================================================================

# 📌 상수 설정
DATASET_NAME = "eddmpython/dartlab-data"
SPLIT_NAME = "train"  # 데이터를 로드할 특정 분할(Split)을 지정합니다.
SAMPLE_COUNT = 10     # 분석할 샘플의 개수 (너무 많으면 시간이 오래 걸리니 10개로 제한!)

print("========================================================")
print("✨ 튜터가 알려주는 AI 데이터 분석 실습을 시작합니다! ✨")
print("========================================================")

# 1. 데이터 로드 (스트리밍 모드 사용)
# streaming=True를 사용하면 데이터셋 전체를 메모리에 올리지 않아 메모리 걱정이 없답니다!
print(f"\n[Step 1/3] 데이터셋 '{DATASET_NAME}'을(를) 스트리밍 모드로 로드합니다...")
try:
    # 데이터를 로드할 때, 필요한 분할과 스트리밍 옵션을 반드시 지정해 주세요!
    full_dataset = load_dataset(DATASET_NAME, split=SPLIT_NAME, streaming=True)
    print("✅ 데이터셋 로드 성공! 이제 분석을 시작할 준비가 되었습니다.")
except Exception as e:
    print(f"❌ 데이터셋 로드에 실패했습니다. 연결 상태를 확인하거나 데이터셋 이름을 다시 확인해주세요. 에러: {e}")
    exit()

# 2. 샘플링 및 이터레이터 준비 (가장 중요한 패턴!)
# 메모리 효율을 위해 전체 데이터셋을 다 읽지 않고, 필요한 만큼만 'take'해서 반복자(iterator)를 만듭니다.
print(f"\n[Step 2/3] 데이터셋에서 상위 {SAMPLE_COUNT}개의 샘플만 골라옵니다...")
try:
    # 💡 핵심 패턴: take()와 iter()를 사용하여 메모리 효율성을 극대화합니다.
    sampled_dataset_iterator: Iterator = iter(full_dataset.take(SAMPLE_COUNT))
    print("✅ 샘플 데이터 이터레이터 준비 완료!")
except Exception as e:
    print(f"❌ 샘플링 중 오류가 발생했습니다: {e}")
    exit()

# 3. 실습: 공시문 핵심 키워드 빈도 분석 (Text Classification 맛보기)
# 금융 공시 데이터는 워낙 방대해서, 모든 내용을 읽기는 어려워요.
# 그래서 상위 샘플의 텍스트 내용을 보고 특정 단어들이 얼마나 자주 나왔는지 '개수 세기' 실습을 해볼 거예요!
print("\n" + "=" * 60)
print("🔬 [Step 3/3] 분석 실습: 상위 샘플에서 핵심 키워드 빈도 탐색")
print("💡 우리는 10개 샘플에서 '매출', '수익', '계약' 단어의 등장 빈도를 계산해봅니다.")
print("=" * 60)

# 📌 분석 키워드 설정 (우리가 찾아보고 싶은 중요한 단어들)
TARGET_KEYWORDS = ["매출", "수익", "계약", "revenue", "profit"]
keyword_counts = {keyword: 0 for keyword in TARGET_KEYWORDS}
processed_sample_count = 0

# 이터레이터(iterator)를 순회하며 샘플을 하나씩 확인합니다.
for i, sample in enumerate(sampled_dataset_iterator):
    # 한 샘플의 내용을 모두 연결하여 분석할 큰 텍스트를 만듭니다.
    # (편의상, 데이터를 담고 있는 모든 문자열 형태의 필드 내용을 임시로 합칩니다.)
    full_text = ""
    for key, value in sample.items():
        # 'features' 키를 가진 필드들은 건너뛰고, 값이 문자열인 필드만 사용합니다.
        if isinstance(value, str):
            full_text += value + " "
    
    # 분석할 텍스트를 소문자 및 공백으로 통일하여 정제합니다.
    cleaned_text = full_text.lower()
    
    # 🔍 키워드 카운팅 시작
    for keyword in TARGET_KEYWORDS:
        # 1. 대소문자 구분 없이 검색하기 위해 키워드를 소문자로 변환합니다.
        search_term = keyword.lower()
        # 2. 텍스트 내에서 해당 키워드가 나타나는 횟수를 계산합니다.
        count = cleaned_text.count(search_term)
        keyword_counts[keyword] += count

    processed_sample_count += 1

# 4. 최종 결과 출력 (튜터의 멋진 마무리!)
print("\n✨ 분석 결과 보고서 (Top {}개 샘플 기준) ✨".format(processed_sample_count))
print("-" * 40)

for keyword, count in keyword_counts.items():
    # 키워드가 발견되었는지 여부를 친절하게 알려줍니다!
    if count > 0:
        print(f"🔎 '{keyword}' 키워드 총 등장 횟수: {count} 회")
    else:
        print(f"🔎 '{keyword}' 키워드는 해당 샘플에서 발견되지 않았습니다.")

print("\n========================================================")
print("🎉 축하합니다! 기본적인 AI 데이터 분석 파이프라인을 완주했습니다.")
print("👏 이 코드는 데이터 로드 -> 샘플링 -> 텍스트 정제 -> 키워드 분석의 과정을 거쳤어요.")
print("다음에는 이 패턴을 활용해 모델을 학습시키면 멋진 AI 전문가가 될 수 있을 거예요!")
print("========================================================\n")
```

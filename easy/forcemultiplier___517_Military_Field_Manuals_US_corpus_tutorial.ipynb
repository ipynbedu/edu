```
import datasets
import re
from collections import Counter

# 📚 [튜터 코딩 가이드]
# 안녕하세요, 예비 AI 개발자님! 💪
# 저희가 오늘 실습할 데이터셋은 '517_Military_Field_Manuals_US_corpus'입니다.
# 이름만 들어도 느껴지죠? 전 세계 군대와 관련된 방대한 '필드 매뉴얼'들이 담겨 있어요. 📰📜
# 즉, 수십 년간의 군사 기록, 기술 설명, 작전 요령 등이 텍스트 형태로 보존된 '초대형 지식 저장소' 같은 데이터셋이랍니다!
# 이 데이터를 분석해서 숨겨진 지식 패턴을 찾아내는 것이 우리가 할 일이에요! 😉

# =====================================================================================
# 💾 데이터셋 기본 정보 (필독!)
# 데이터셋 제목: 517_Military_Field_Manuals_US_corpus
# 의미: 미국 군대에서 사용되던 다양한 종류의 현장 매뉴얼 및 문헌 자료 모음.
# 용도: 매우 방대한 양의 텍스트를 다루는 NLP(자연어 처리) 모델을 훈련시키기 아주 좋은 데이터셋입니다.
# =====================================================================================

# 🚀 실습 환경 설정 (constaints 충족을 위한 상수 정의)
DATASET_NAME = "forcemultiplier/517_Military_Field_Manuals_US_corpus"
SAMPLE_COUNT = 10  # 스트리밍 학습의 효율성을 위해 최대 10개의 샘플만 사용해봅시다!

def run_military_manuals_analyzer():
    """
    제공된 군사 매뉴얼 데이터셋을 스트리밍 방식으로 로드하고,
    초보자가 흥미를 느낄 수 있는 간단한 텍스트 분석 실습을 수행합니다.
    """
    print("=" * 70)
    print(f"✨ [미션 시작!] '{DATASET_NAME}' 텍스트 분석기 가동! ✨")
    print("====================================================================")

    # 📌 Step 1: 데이터셋 로드 (Streaming 모드 사용!)
    # 데이터셋이 너무 크기 때문에, 'streaming=True'를 사용해 필요할 때만 데이터를 가져올 거예요.
    # 시간이 절약되고 메모리 사용량도 확 줄어드는 마법 같은 방법이랍니다!
    try:
        # split='train'을 명시하여 로드합니다.
        dataset = datasets.load_dataset(DATASET_NAME, split='train', streaming=True)
        print(f"✅ 데이터셋 로드 완료! (스트리밍 모드, split='train')")
    except Exception as e:
        print(f"🚨 데이터셋 로드 중 오류 발생: {e}")
        print("실행 환경을 확인해주세요.")
        return

    # 📌 Step 2: 샘플러 준비 및 Iterating (핵심 패턴!)
    # 전체 데이터셋을 list()로 불러오는 것은 불가능하죠! 대신 .take()와 iter()를 써서 상위 N개만 루프할 거예요.
    print(f"\n📖 Top {SAMPLE_COUNT}개의 샘플만 가지고 텍스트 분석을 시작합니다...")
    
    # Constraint 9: .take()를 사용한 스트리밍 샘플링
    sampled_dataset_iterator = iter(dataset.take(SAMPLE_COUNT))

    # 분석 결과를 저장할 변수들
    total_words_count = 0
    sample_texts = []
    
    # 📚 분석 결과를 저장하기 위한 데이터 구조
    all_keywords = []
    
    # 💡 Constraint 1-1: 'features' 키를 직접 건너뛰고, 데이터 본체에 있는 필드를 사용합니다.
    
    # 📝 Step 3: 루프를 돌며 샘플 데이터를 하나씩 분석합니다.
    sample_count = 0
    for sample_data in sampled_dataset_iterator:
        sample_count += 1
        
        # 실제 텍스트 데이터가 담겨 있는 필드 이름을 사용합니다.
        # (이 데이터셋의 구조를 가정하여 'content' 필드를 사용합니다.)
        text_content = sample_data.get('content', '') 
        
        if not text_content:
            # 내용이 비어있으면 다음 샘플로 넘어갑니다.
            continue
            
        # ------------------------------------------------
        # 🧠 [창의적 실습 1] 문서의 길이를 재어보자! (문서 분석 기초)
        # ------------------------------------------------
        word_count = len(text_content.split())
        total_words_count += word_count
        sample_texts.append(text_content)
        
        # ------------------------------------------------
        # 🔍 [창의적 실습 2] 핵심 키워드 빈도수 측정! (자연어 처리 기초)
        # 매뉴얼은 특정 단어(키워드)가 반복됩니다. 어떤 단어가 가장 많이 나올까요?
        # ------------------------------------------------
        
        # 1. 구두점을 제거하고 소문자로 변환 (전처리 과정)
        clean_text = re.sub(r'[^\w\s]', '', text_content).lower()
        # 2. 단어 리스트로 분리
        words = clean_text.split()
        
        # 3. 모든 단어를 키워드 리스트에 추가
        all_keywords.extend(words)
        
        # 너무 길어서 분석이 어려우니, 일단 가장 대표적인 텍스트만 출력해볼게요.
        if sample_count <= 3:
             print(f"--- [{sample_count}번째 샘플] ---")
             print(f"📄 내용 일부: {text_content[:100]}...")
             print(f"🔢 단어 수: {word_count}개")


    # 🚀 Step 4: 분석 결과 종합 및 출력
    print("\n" + "=" * 70)
    print("🎉 분석 미션 완료! 최종 결과 보고서입니다. 🎉")
    print("=" * 70)
    
    print(f"\n✨ 분석 대상: 총 {sample_count}개의 매뉴얼 샘플")
    print(f"📚 총 단어 수 (전체 샘플 합): 약 {total_words_count:,}개")

    # 가장 많이 등장한 키워드 TOP 5 분석
    keyword_counts = Counter(all_keywords)
    most_common_keywords = keyword_counts.most_common(5)
    
    print("\n🔑 [통찰력 발휘] 이 매뉴얼들이 가장 많이 다룬 상위 5개 키워드:")
    for keyword, count in most_common_keywords:
        print(f"  - '{keyword}' (횟수: {count})")

    print("\n🏆 [튜터 코멘트]")
    print("축하합니다! 복잡한 스트리밍 데이터셋을 성공적으로 탐색했습니다!")
    print("단순히 데이터를 읽는 것을 넘어, '단어'라는 패턴을 뽑아내는 과정(Tokenization/Frequency Analysis)까지 성공했어요.")
    print("다음 단계에서는 '이 키워드들이 어떤 주제와 관련되어 있는지'를 찾아보는 NMT(신경망 기계 번역)를 배워봅시다! 👍")


# 메인 실행 함수 호출
run_military_manuals_analyzer()
```

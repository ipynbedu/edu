```
import random
from datasets import load_dataset
import os

# =============================================================================
# ✨ 친절한 튜터 모드 ON! ✨
# 📝 데이터셋 이름: Quadyun/Korean_SAT_MATH
# 📚 데이터셋 설명: 대한민국 대학수학능력시험(SAT) 기출 수학 문제 데이터입니다.
# 🎓 학습 목표: 이 코드를 통해 대규모 시험 문제 데이터셋을 효율적으로 로드하고,
#          데이터의 구조적 특징을 분석하여 간단한 AI 분석 로직을 구현하는 것을 배워봅시다.
# 💡 오늘 배우는 핵심: 대규모 데이터를 메모리 효율적으로 처리하는 '스트리밍' 기술과
#                    데이터 패턴을 추출하는 기초적인 AI 분석 기법입니다!
# =============================================================================

# 📌 데이터셋 ID 설정
DATASET_NAME = "Quadyun/Korean_SAT_MATH"
SAMPLE_COUNT = 10 # 처음에는 10개 샘플만 볼 거예요! (빠른 학습을 위해)

# --- 1. 데이터 로드 준비 ---

# 이전에 로드했던 dataset 객체를 초기화합니다.
dataset = None
selected_split = 'test'

print("======================================================================")
print(f"📚 데이터셋 로드 시작: {DATASET_NAME} - {selected_split} 스플릿")
print("======================================================================")

# 🚀 스트리밍(Streaming) 모드 시도 (가장 빠르고 메모리 효율적!)
try:
    # 데이터를 스트리밍 모드로 최대한 빨리 로드해 봅시다.
    dataset = load_dataset(DATASET_NAME, split=selected_split, streaming=True)
    print("✨ [성공] 데이터를 스트리밍 모드(Streaming)로 성공적으로 로드했습니다. 메모리 절약 만점!")
except Exception as e:
    print(f"⚠️ [경고] 스트리밍 로드에 실패했습니다 ({e}). 메모리를 많이 쓰더라도 일부 데이터를 직접 다운로드하겠습니다.")
    # 스트리밍이 안 되면, 소량만 다운로드해서 일반 Dataset 객체로 만듭니다.
    try:
        dataset = load_dataset(DATASET_NAME, split=selected_split)
        print("✨ [대체] 일반 데이터셋(Dataset) 모드로 로드했습니다. 이제 마음껏 데이터를 만져봐요!")
    except Exception as e_fallback:
        print(f"❌ [실패] 데이터셋 로드에 실패했습니다. 인터넷 연결을 확인해주세요. 오류: {e_fallback}")
        exit()

# --- 2. 샘플링 및 데이터 전처리 (Beginner's Toolkit) ---

# ✅ 스트리밍/일반 데이터셋에 관계없이, 안전하게 샘플만 가져오는 마법!
print("\n======================================================================")
print(f"💡 샘플 데이터 {SAMPLE_COUNT}개 준비 중...")

# 만약 dataset 객체에 take() 메소드가 있다면 (스트리밍 모드일 가능성이 높음)
if hasattr(dataset, "take"):
    # .take()를 사용해 상위 SAMPLE_COUNT 개만 가져와서 리스트로 만듭니다.
    # 이 방식이 '스트리밍' 데이터셋에서 가장 안전하고 빠릅니다.
    sample_dataset_list = list(dataset.take(SAMPLE_COUNT))
else:
    # 일반 Dataset 객체인 경우
    sample_dataset_list = list(dataset.select(range(min(SAMPLE_COUNT, 100)))) # 최대 100개로 제한하여 안전하게 샘플링

sample_data_list = sample_dataset_list

print(f"✅ 준비 완료! 총 {len(sample_data_list)}개의 샘플을 분석할 수 있게 되었습니다.")


# --- 3. 데이터 분석 실습: 수학 문제의 구조 파헤치기 (The Data Explorer) ---

print("\n======================================================================")
print("🧐 Step 3: 데이터 분석 - 문제 패턴 탐색하기 (Pattern Discovery)")
print("======================================================================")

# 🎨 분석 1: 문제 이미지 의존도 분석 (매우 중요!)
print("\n[🖼️ 1. 이미지 의존도 분석]")
image_count = sum(1 for sample in sample_data_list if sample['has_image'])
total_count = len(sample_data_list)
print(f"총 {total_count}개 샘플 중, 이미지가 포함된 문제는 {image_count}개 입니다.")
print(f"👉 이미지 의존 비율: {(image_count / total_count * 100):.2f}%")

if image_count > 0:
    print("💡 튜터 Tip: 이 비율이 높다면, 이 문제는 '이미지 처리 모델(Vision Model)'이 필수라는 뜻이에요!")
else:
    print("💡 튜터 Tip: 현재 샘플에는 이미지가 없습니다. 다른 샘플을 로드해보세요!")


# 🎨 분석 2: 오답률/정답률의 평균 분석 (기초 통계)
print("\n[💯 2. 점수 및 정답 분석 (Score Analysis)]")
total_score = 0
correct_predictions = 0
for sample in sample_data_list:
    total_score += sample['score']
    if sample['answer'] == 'A': # 가상의 정답 확인 로직 (answer 필드 확인)
        correct_predictions += 1

print(f"✅ 분석한 샘플들의 총 점수 합계: {total_score} 점")
print(f"🌟 샘플 중 'A'를 정답으로 가정하고 분석한 횟수: {correct_predictions} 회")
print("📈 다음 단계: 이 점수와 정답 패턴을 이용해 '가장 쉬운 문제'를 분류하는 모델을 만들 수 있어요!")


# --- 4. 창의적인 AI 활용 예시: LLM 프롬프트 생성 시뮬레이션 (The Creative Jump) ---

print("\n======================================================================")
print("🤖 Step 4: AI 활용 시뮬레이션 - 프롬프트 엔지니어링")
print("======================================================================")

if sample_data_list:
    # 첫 번째 샘플을 예시로 사용합니다.
    example_sample = sample_data_list[0]
    
    print(f"🔍 분석 대상 문제 (ID: {example_sample['id']}):")
    print("-" * 50)
    print(f"  [문제 내용]: {example_sample['question'][:50]}...")
    
    # 문제의 핵심 정보만 추출해서 프롬프트를 만듭니다.
    prompt_template = f"""
    [Context]: 당신은 대한민국 수능 전문 튜터입니다.
    [Question]: {example_sample['question']}
    [Choices]: {', '.join(example_sample['choices'])}
    [Goal]: 위의 질문에 대한 논리적 풀이 과정을 상세하게 설명하고, 최종 정답만 한글로 제시하세요.
    """
    
    print("\n💡 AI 모델에게 던져줄 최적의 명령어(Prompt) 구조:")
    print("----------------------------------------------------------------------")
    # f-string을 사용하여 코드가 잘 보이게 출력합니다.
    print(prompt_template.strip())
    print("----------------------------------------------------------------------")
    print("🥳 축하해요! 데이터를 분석하고, AI가 이해할 수 있는 언어로 명령을 설계하는 과정을 완료했습니다!")
else:
    print("🤷‍♀️ 샘플 데이터가 준비되지 않아 시뮬레이션을 건너뜁니다.")

print("\n======================================================================")
print("🎉 모든 실습 과정이 완료되었습니다! 데이터에 대한 이해도가 확 높아지셨을 거예요!")
print("🎉 다음에는 실제로 이 데이터를 이용해 '분류 모델'을 만들어 봅시다!")
print("======================================================================")
```

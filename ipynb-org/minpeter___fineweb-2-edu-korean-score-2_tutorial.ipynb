```
import random
from datasets import load_dataset, get_dataset_config_names

# 📚 데이터셋 정보
# 데이터셋 이름: minpeter/fineweb-2-edu-korean-score-2
# 의미: 웹에서 스크랩한 한국어 교육/학습 콘텐츠 모음집입니다.
# 특징: 각 글의 품질 점수(score), 언어 정보(language_script), 전문 텍스트(text) 등 풍부한 메타데이터를 가지고 있어,
# 데이터 전처리, 필터링, 분석 연습에 매우 적합합니다!
DATASET_NAME = "minpeter/fineweb-2-edu-korean-score-2"
SAMPLE_COUNT = 5  # 실습에서 사용할 샘플 개수 (너무 많으면 시간이 오래 걸려요!)

# --------------------------------------------------------------------------
# 🚀 튜터: "환영합니다, 미래의 AI 개발자님! 우리는 데이터 속 숨겨진 보석을 찾는 탐험가가 될 거예요.
# 이 데이터셋은 엄청나게 큰 온라인 학습 자료 창고예요. 지금부터 가장 신뢰성 높은 '꿀팁'만 추려내는 작업을 해봅시다!"
# --------------------------------------------------------------------------

print("=" * 80)
print(f"✨ ✨ 초급 AI 실습: 고품질 학습 콘텐츠 필터링 엔진 개발 ✨ ✨")
print("=" * 80)

dataset = None
sample_data_list = []

# 1. 데이터 로딩 시도: 스트리밍 방식 (최신 트렌드를 따라잡기!)
try:
    print("💡 STEP 1/4: 데이터셋을 메모리 효율적인 '스트리밍' 모드로 불러오기를 시도합니다...")
    dataset = load_dataset(DATASET_NAME, split='train', streaming=True)
    print("✅ 스트리밍 로드 성공! (메모리 걱정 없이 빠르게 샘플링할 수 있어요)")

except Exception as e:
    # 스트리밍 모드가 너무 크거나 복잡하여 실패할 수 있습니다. 이럴 땐 일반 다운로드 방식으로 전환합니다.
    print(f"⚠️ 경고: 스트리밍 로드 실패 또는 권한 문제 ({e.__class__.__name__}).")
    print("💾 일반 다운로드 모드로 전환하여 소량의 데이터를 불러올게요. (시간이 조금 더 걸릴 수 있어요)")
    try:
        # 소량만 다운로드하여 안전하게 실행하도록 설정
        dataset = load_dataset(DATASET_NAME, split='train', streaming=False)
    except Exception as e_fallback:
        print(f"❌ 치명적인 오류 발생! 데이터셋을 로드할 수 없습니다. 오류: {e_fallback}")
        exit()

# 2. 데이터 전처리 및 샘플 확보
# 스트리밍 모드와 일반 모드 모두 안전하게 샘플을 얻는 패턴을 사용합니다.
if hasattr(dataset, "take"):
    # .take()가 존재하면 스트리밍 데이터셋(IterableDataset)
    print(f"\n💡 STEP 2/4: {SAMPLE_COUNT}개의 샘플 데이터를 확보하고 분석을 준비합니다...")
    sample_iterator = dataset.take(SAMPLE_COUNT)
    sample_data_list = list(sample_iterator)
else:
    # 일반 데이터셋 (Dataset)
    sample_data_list = list(dataset.select(range(SAMPLE_COUNT)))

# 3. 창의적 실습: '고품질 한국어 학습 자료 선별' 필터링
print("\n💡 STEP 3/4: [🔥 핵심 분석 구간] 고품질 한국어 콘텐츠 필터링 엔진을 가동합니다.")
print("   우리의 목표: 1. 언어 스크립트가 한글(Hangul)이고, 2. 품질 점수(score)가 높은 글만 골라내기!")

high_quality_samples = []
for i, sample in enumerate(sample_data_list):
    # 🎯 필터링 조건 설정
    is_korean = sample.get('language_script') == 'Hangul'
    is_high_score = sample.get('score', 0.0) >= 0.5

    # 필터 통과 여부 판단
    if is_korean and is_high_score:
        high_quality_samples.append(sample)
    else:
        pass # 필터링을 통과하지 못하면 무시

# 4. 결과 시각화 및 활용 (LLM Prompt Generator처럼 사용)
print("\n✨✨ STEP 4/4: 분석 결과 보고서 출력 (만점 자료로 AI 튜터링 프롬프트 만들기!) ✨✨")
print("-" * 60)

if high_quality_samples:
    print(f"🏆 성공! 총 {len(high_quality_samples)}개의 고품질 한국어 자료를 선별했습니다.")
    print("   이 자료들을 바탕으로 '다음 질문에 답해줘'라는 완벽한 프롬프트를 만들 수 있어요!")
    
    for i, sample in enumerate(high_quality_samples):
        score = sample.get('score', 0.0)
        text = sample.get('text', 'N/A')
        date = sample.get('date', 'Unknown Date')
        
        print(f"\n[📚 자료 {i+1} - ✨ 추천 점수: {score:.4f} (A+급)]")
        print(f"   📅 출처일: {date}")
        
        # ✍️ 분석 결과 출력: 텍스트의 일부를 가져와서 예시로 보여줍니다.
        print("   [📝 원문 미리보기]")
        preview = text[:100] + "..." if len(text) > 100 else text
        print(f"   -> \"{preview}\"")

        # 🤖 AI 활용 제안 (가장 창의적인 부분)
        print("\n   ✨ AI 튜터 활용 제안 (Prompt Engineering):")
        print(f"   → [역할 부여]: \"너는 {sample.get('language', '한국어')} 전문 교사야.\"")
        print(f"   → [명령]: \"위의 지문 내용을 바탕으로, 1. 핵심 키워드 3가지 요약해 주고, 2. 이 내용을 초등학교 3학년 눈높이에 맞춰 비유를 들어 설명해줘.\"")
        print("   (이런 식으로 메타데이터와 텍스트를 조합하면 최고의 AI 질문이 됩니다!)")
else:
    print("😞 아쉽게도, 테스트 샘플에서는 높은 점수와 한글 스크립트를 모두 만족하는 자료를 찾지 못했어요.")
    print("   👉 이럴 땐 필터링 기준(score >= 0.5)을 조금 더 낮춰서 다시 시도해 보세요!")

print("\n" + "=" * 80)
print("🎉 축하합니다! 이제 데이터에서 의미 있는 정보를 추출하는 능력을 갖추셨어요. 정말 잘하셨습니다! 💪")
print("=" * 80)
```

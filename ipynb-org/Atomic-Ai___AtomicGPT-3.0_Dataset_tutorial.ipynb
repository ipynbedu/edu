```
# 🤖✨ 환영합니다! 초보 개발자를 위한 AI 코딩 튜터입니다! ✨🤖
# 이 스크립트는 방대한 법률 및 기술 문서로 이루어진 'AtomicGPT-3.0_Dataset'을
# 활용하여 기본적인 텍스트 데이터 처리와 AI가 무엇을 하는지 '체험'해 보는 실습입니다.
# 📄 데이터셋: Atomic-Ai/AtomicGPT-3.0_Dataset (법률/규정 텍스트)
# 💡 목표: 데이터셋의 규모와 내용 구조를 파악하고, NLP(자연어 처리)의 기본 과정을 이해합니다.

import random
from datasets import load_dataset

# --- 📚 상수 설정 ---
DATASET_NAME = "Atomic-Ai/AtomicGPT-3.0_Dataset"
SAMPLE_COUNT = 100 # 🤔 전체 데이터를 다 볼 순 없으니까, 일단 상위 100개만 맛보기로 볼 거예요!

print("=============================================================")
print(f"✨ 데이터셋 로딩을 시작합니다: {DATASET_NAME}")
print("=============================================================")

# 📌 STEP 1: 데이터 로드 전략 수립 (스트리밍 체크!)
# 데이터셋이 너무 클 경우, 스트리밍(streaming=True)을 사용하면 메모리 효율적입니다.
dataset = None
try:
    # 🚀 스트리밍 모드 시도 (빠르고 효율적!)
    print("▶️ 스트리밍 모드(Streaming=True)로 데이터셋을 로드합니다...")
    # split='train'을 명시하고 스트리밍으로 로드합니다.
    dataset = load_dataset(DATASET_NAME, split='train', streaming=True)
    print("✅ 성공! 데이터셋을 스트리밍 방식으로 로드했습니다. 메모리 걱정 No!")

except Exception as e:
    # 🚨 스트리밍 로드 실패 시 백업 플랜 작동!
    print(f"\n⚠️ [경고] 스트리밍 로드 실패 ({e}). 일반 다운로드 모드로 전환합니다.")
    try:
        # 일반 모드로 소량만 다운로드하여 안전하게 진행합니다.
        dataset = load_dataset(DATASET_NAME, split='train')
        print("✅ 백업 성공! 일반 데이터셋 모드로 로드했습니다. (느리지만 안전!)")
    except Exception as e_fallback:
        print(f"❌ 치명적인 오류 발생! 데이터셋 로드에 실패했습니다: {e_fallback}")
        exit()


# 📌 STEP 2: 데이터 반복 처리 방식 결정 (Dataset vs IterableDataset)
if hasattr(dataset, "take"):
    # .take()가 존재하면 스트리밍 데이터셋(IterableDataset)일 가능성이 높습니다.
    print("\n▶️ 현재 데이터셋은 스트리밍 방식으로 작동합니다. .take()를 사용하여 데이터를 가져올게요.")
    # 데이터셋의 처음 SAMPLE_COUNT 개를 반복 가능한 이터레이터로 만듭니다.
    sampled_dataset_iterator = dataset.take(SAMPLE_COUNT)
else:
    # 일반 데이터셋 (Dataset)인 경우
    print("\n▶️ 현재 데이터셋은 일반(Standard) 데이터셋입니다. .take()를 사용합니다.")
    # list(dataset.take(SAMPLE_COUNT)) 패턴을 사용하여 리스트로 만듭니다.
    list_dataset = list(dataset.take(SAMPLE_COUNT))
    sampled_dataset_iterator = list_dataset


# 📌 STEP 3: 데이터 샘플 추출 (실습용 리스트 만들기)
# 반복문 처리를 위해 iterator를 list로 변환합니다.
sample_data_list = list(sampled_dataset_iterator)

if not sample_data_list:
    print("\n😭 로드된 샘플 데이터가 없습니다. 코드를 다시 확인해주세요!")
    exit()

print(f"\n=============================================================")
print(f"✨ 🌟 [샘플 데이터 분석] 총 {len(sample_data_list)}개의 샘플을 분석합니다! 🌟 ✨")
print("=============================================================")


# ===================================================================
# 🧠 실습 1: 데이터 구조 탐색 및 기본 통계 분석 (Quantitative Analysis)
# ===================================================================
print("\n[📊 실습 1] 샘플 데이터의 텍스트 길이 통계 분석 (Text Length Stats)")
print("----------------------------------------------------------------------")

# 📚 데이터 구조 확인 (우리가 다룰 필드는 'text' 하나뿐이네요!)
sample_texts = [sample['text'] for sample in sample_data_list]

if sample_texts:
    # 🔢 모든 샘플의 텍스트 길이를 저장합니다.
    text_lengths = [len(text) for text in sample_texts if isinstance(text, str)]
    
    if text_lengths:
        # 📈 통계 계산 (평균, 최대, 최소)
        avg_length = sum(text_lengths) / len(text_lengths)
        max_length = max(text_lengths)
        min_length = min(text_lengths)

        print(f"✅ 분석 결과 (총 {len(text_lengths)}개 샘플 기준):")
        print(f"  🔹 평균 텍스트 길이: {avg_length:.2f} 글자 (Wow, 꽤 긴 법률 문서 같아요!)")
        print(f"  🔹 최대 텍스트 길이: {max_length} 글자")
        print(f"  🔹 최소 텍스트 길이: {min_length} 글자")
    else:
        print("⚠️ 텍스트 필드에 분석 가능한 문자열이 없습니다.")


# ===================================================================
# 🤖 실습 2: 간단한 NLP 시뮬레이션 (Prompt Engineering Simulation)
# ===================================================================
print("\n[🔍 실습 2] AI가 문맥을 파악하는 시뮬레이션: 핵심 주제 키워드 검색")
print("----------------------------------------------------------------------")
print("💡 목표: 이 텍스트가 '지적재산권(IP)'에 대한 것인지, 아니면 '사법 절차(Legal Process)'에 대한 것인지 판별해봅시다.")

# 🔎 핵심 키워드 설정 (독일어 환경이므로 독일어와 유사한 주제어를 사용합니다.)
KEYWORDS_IP = ["patent", "trademark", "rechtsverletzung", "geistiges eigentum"] # 지적재산권 관련 독일어 단어
KEYWORDS_LAW = ["gericht", "klage", "verordnung", "rechtsprechung"] # 법원/소송/규정 관련 독일어 단어

print(f"🔎 [키워드 세트]: IP 키워드 ({KEYWORDS_IP[0]}...), 법률 키워드 ({KEYWORDS_LAW[0]}...)")

ip_count = 0
law_count = 0
total_processed = 0

for sample in sample_data_list:
    text = sample['text']
    if not isinstance(text, str):
        continue
    
    total_processed += 1
    
    # 🏷️ 단순 카운팅 기반 분류 시뮬레이션 (LLM의 기초 원리 체험!)
    found_ip = any(keyword in text.lower() for keyword in KEYWORDS_IP)
    found_law = any(keyword in text.lower() for keyword in KEYWORDS_LAW)

    if found_ip:
        ip_count += 1
    if found_law:
        law_count += 1

# 📊 최종 결과 요약
print("\n🌟 분석 요약 (전체 샘플 중):")
print(f"  📚 총 처리된 샘플 수: {total_processed}개")
print(f"  🛡️ 지적재산권(IP) 관련 주제가 포함된 샘플 수: {ip_count}개")
print(f"  ⚖️ 사법 절차(Law) 관련 주제가 포함된 샘플 수: {law_count}개")

if total_processed > 0:
    ip_ratio = (ip_count / total_processed) * 100
    law_ratio = (law_count / total_processed) * 100
    print(f"\n✨ 해석: 샘플 중 IP 관련 비중은 약 {ip_ratio:.1f}%, 법률 절차 비중은 약 {law_ratio:.1f}% 입니다.")
    print("\n🎉 축하합니다! 당신은 이제 데이터의 패턴을 파악하는 기초적인 AI 분석 능력을 갖추었습니다!")
```

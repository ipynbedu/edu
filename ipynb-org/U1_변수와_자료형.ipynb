## 1. 변수(Variable)란?

변수는 **값에 이름을 붙여서 저장해두는 것**입니다. 상자에 이름표를 붙이는 것과 같아요.

실제 AICE Associate 시험 문제 2번은 와인 데이터를 `data`라는 변수에 담으라고 요구합니다:

```python
data = pd.read_csv('./wine_quality_data.csv')
```

지금은 `pd.read_csv(...)`가 뭔지 몰라도 됩니다. 핵심은 이 구조입니다 — **`변수이름 = 값`**.

이 데이터의 두 번째 행(와인 하나)에는 실제로 이런 값들이 들어 있습니다.

| 컬럼명 | 뜻 | 실제 값 |
|---|---|---|
| `E_country` | 생산국가 | `Portugal` |
| `E_price` | 가격 | `15.0` |
| `E_FE_vintage` | 포도 수확연도 | `2011` |
| `Grade` | 품질등급(Pass/Fail) | `Fail` |

이 값들을 파이썬 변수로 옮겨서 확인해봅시다.

````python
E_country = "Portugal"
print(E_country)
````

`print()`는 괄호 안에 든 값을 화면에 보여달라는 명령입니다. 변수를 만들기만 하고 `print()`를 안 하면, 값이 상자 안에 들어있긴 해도 화면엔 아무것도 안 보입니다.

**주의할 점 하나.** `=`는 수학에서 배운 "같다"가 아닙니다. 파이썬에서 `=`는 **"오른쪽 값을 왼쪽 이름에 넣어라"**는 뜻입니다. ("같다"를 뜻하는 기호는 나중에 U3에서 따로 배웁니다 — 지금은 헷갈리지 않게 `=`은 무조건 "넣어라"로만 기억하세요.)

## 2. 자료형(Data Type) — 값에는 "종류"가 있다

| 자료형 | 이름 | 위 표에서의 예 |
|---|---|---|
| `int` | 정수 (소수점 없는 숫자) | `E_FE_vintage`(수확연도) → `2011` |
| `float` | 실수 (소수점 있는 숫자) | `E_price`(가격) → `15.0` |
| `str` | 문자열 (글자) — 반드시 따옴표로 감쌈 | `E_country`(국가) → `"Portugal"` |
| `bool` | 불리언 (참/거짓, 둘 중 하나만) | 이 와인의 등급이 Fail인지 → `True`/`False` |

**중요한 함정 하나.** `2011`(숫자)과 `"2011"`(문자열)은 겉보기엔 비슷해 보여도 완전히 다른 것입니다. `"2011"`은 계산할 수 없는 그냥 글자 네 개일 뿐입니다.

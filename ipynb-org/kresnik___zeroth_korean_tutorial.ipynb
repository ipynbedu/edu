```
import random
import numpy as np
from datasets import load_dataset
import os

# ======================================================================
# 💖 AI 코딩 튜터가 준비한 초급 파이썬 스크립트: 한국어 음성 데이터 탐험!
# 💖 제목: Zeroth-Korean 데이터셋을 활용한 '화자 및 주제 분석' 시뮬레이터
# 💖 데이터셋 의미: 한국어 ASR 연구를 위한 음성-텍스트 쌍 데이터셋입니다.
# 💖 목적: 실제 대용량 음성 데이터셋에서 어떻게 소량의 샘플만으로 의미있는 인사이트를 도출하는지 연습합니다.
# ======================================================================

# 상수 설정 (튜너가 정해준 값들)
DATASET_NAME = "kresnik/zeroth_korean"
SAMPLE_COUNT = 10  # ✨ 초보자 실습을 위해 딱 10개의 샘플만 사용해요!

print("✨ 안녕하세요! 여러분의 코딩 튜터입니다! 오늘은 엄청나게 큰 음성 데이터셋을 다뤄볼 거예요.")
print("📚 이 데이터셋은 한국어 ASR (자동 음성 인식) 연구에 쓰인 보물상자 같은 자료예요!")
print("-" * 50)

# 1. 데이터 로드: 스트리밍 vs. 다운로드 전략 (실패에 대비하는 꼼수!)
dataset = None
try:
    # 🏃‍♀️ 최우선 전략: 메모리 효율이 최고인 스트리밍 모드를 시도합니다.
    print("✅ 1단계: Streaming 모드를 시도합니다... (가장 빠르고 가벼운 방법!)")
    dataset = load_dataset(DATASET_NAME, split='train', streaming=True)
    print("🎉 성공! 데이터셋을 스트리밍 모드로 성공적으로 로드했어요. 대용량 데이터도 메모리 걱정 없이 돌릴 수 있어요!")

except Exception as e:
    # 🚨 예외 처리: 스트리밍이 실패할 경우, 작은 분량만 다운로드하여 강행합니다.
    print(f"⚠️ 경고: 스트리밍 로드 중 오류 발생 ({e.__class__.__name__}). ")
    print("📉 대안으로, 테스트 세트 중 소량만 다운로드하여 진행할게요. 너무 걱정 마세요!")
    try:
        dataset = load_dataset(DATASET_NAME, split='test', streaming=False)
    except Exception as fallback_e:
        print(f"❌ 모든 로드 시도가 실패했습니다. 데이터셋 ID를 확인해주세요. ({fallback_e})")
        exit()

# 2. 샘플 데이터 추출 (무한한 데이터 속에서 필요한 조각만 쏙!)
print("\n🔎 2단계: 전체 데이터셋 대신, 딱 필요한 샘플 10개만 뽑아와요.")
# 🚀 제약 조건: 'len()'이나 'select'를 쓰지 않고, 'take()'를 사용해요!
sampled_dataset_iterator = dataset.take(SAMPLE_COUNT)
sample_data_list = list(sampled_dataset_iterator)

# 💡 분석 결과 출력
print(f"📊 데이터 분석: 로드된 전체 데이터셋은 {DATASET_NAME} 입니다.")
print(f"✅ 추출 샘플 수: {len(sample_data_list)}개 (훌륭해요! 관리하기 딱 좋은 크기예요!)")
print("-" * 50)


# 3. 창의적 실습: 화자별 대화 주제 추론 및 AI 프롬프트 생성 시뮬레이션
print("\n🌟 3단계: '화자 및 주제 프로파일링' 실습을 시작합니다!")
print("Goal: 각 샘플의 텍스트를 보고, 누가(Speaker) 무슨 말을 했는지 파악하여, 다음 대화를 위한 AI 프롬프트를 만들어 볼 거예요.")
print("======================================================================================")

# 💖 데이터 순회 시작! (튜터의 꿀팁: list(iterator)를 사용하면 깔끔하게 반복할 수 있어요.)
for i, sample in enumerate(sample_data_list):
    text = sample['text']
    speaker_id = sample['speaker_id']
    chapter_id = sample['chapter_id']
    
    # 🤓 메타데이터 분석: 화자 ID와 챕터 ID를 활용합니다.
    if not text:
        print(f"[{i+1}번] ℹ️ 텍스트가 비어있네요. 건너뜁니다.")
        continue

    # 💡 분석 로직: 텍스트가 어떤 키워드를 포함하는지 간단히 확인하여 주제를 추론합니다.
    theme = "✅ 일반적인 일상 대화"
    if '한국' in text or '조국' in text:
        theme = "🇰🇷 한국 관련 문화/역사 주제"
    elif '기술' in text or 'ai' in text:
        theme = "🤖 기술 및 미래 기술 토론"
    elif '밥' in text or '식사' in text:
        theme = "🍔 식사 및 음식 관련 대화"
    
    # 🚀 AI 프롬프트 생성 시뮬레이션 (이게 바로 AI 활용의 꽃!)
    prompt = f"""
    [역할 설정]: 당신은 전문 대화 분석가입니다.
    [현재 대화 정보]: 화자 ID {speaker_id}가 '제 {chapter_id} 챕터'에서 "{text[:40]}..." 라고 말했습니다.
    [주제]: {theme}
    [요청]: 이 주제를 심화하기 위한, 반대 의견이나 질문 3가지를 영어로 제시해주세요.
    """
    
    # ✨ 결과 출력 (튜터의 위트 넘치는 코멘트와 함께!)
    print(f"\n--- 💡 Sample {i+1} / Chapter {chapter_id} ---")
    print(f"   🗣️ 화자 ID: {speaker_id} | 📖 추정 주제: {theme}")
    print(f"   💬 원문 텍스트: '{text}'")
    print("\n   🤖 [자동 생성된 LLM 프롬프트 예시]:")
    # 실제 LLM에 넣기 전에, 이 프롬프트가 어떻게 구조화되어야 하는지 눈으로 확인하는 과정이 중요해요!
    print("   >>>" + "\n   >>>".join(prompt.split('\n')))

print("\n======================================================================================")
print("✨ 🎉 축하합니다! 초급 AI 데이터 분석 실습을 완벽하게 마쳤습니다!")
print("🚀 핵심 학습 포인트:")
print("1. **Streaming Dataset 처리:** 대규모 데이터셋을 메모리 걱정 없이 효율적으로 다루는 방법을 익혔어요.")
print("2. **Metadata 결합:** 음성 데이터(Audio)와 텍스트(Text), 그리고 화자 정보(Speaker ID)라는 3가지 종류의 정보를 결합하여 분석하는 능력을 길렀어요.")
print("3. **실용적인 추론:** 단순 데이터 출력에 그치지 않고, '이 데이터로 어떤 AI 서비스를 만들 수 있을까?'라는 질문을 던져 프롬프트까지 짜보는 상상력을 발휘했어요!")
```

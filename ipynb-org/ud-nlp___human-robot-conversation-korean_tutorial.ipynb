```
# ==============================================================================
# ✨ Human-Robot 대화 데이터셋 탐험기: 음성 인식 기초 실습 ✨
# 📚 제목: ud-nlp/human-robot-conversation-korean (한국어 인간-로봇 대화 데이터셋)
# 💬 의미: AI와 인간 사이의 실제 대화 내용을 녹음한 오디오 데이터셋입니다.
# 🤖 설명: 이 데이터셋은 'STT(Speech-to-Text)', 즉 음성 인식(ASR) 기술을 훈련하는 데 사용됩니다.
#          우리는 이 데이터를 분석하여, AI가 어떻게 사람의 말을 인식하고 대화 흐름을 이해하는지
#          파이썬 코드만으로 맛보기 실습을 진행해 볼 거예요. 걱정 마세요, 코딩이 어려워도 괜찮아요! 😊
# ==============================================================================

import random
import numpy as np
from datasets import load_dataset, Dataset

# 💡 튜터의 한 마디: 오늘의 목표는 데이터가 어떻게 구조화되어 있는지, 그리고
# 우리가 어떤 데이터를 꺼내 분석할 수 있는지 '눈으로 확인'하는 거예요!

DATASET_NAME = "ud-nlp/human-robot-conversation-korean"
SPLIT_NAME = "train"
SAMPLE_COUNT = 10

# ------------------------------------------------------------------------------
# 🚀 1단계: 데이터셋 로드하기 (지능형 로딩 패턴 적용!)
# ------------------------------------------------------------------------------

print("=" * 80)
print("🤖 로봇 AI가 당신을 환영합니다! 데이터셋 로드를 시작하겠습니다...")
print("=" * 80)

dataset = None
sample_data_list = []

try:
    # 🟢 1차 시도: 스트리밍 모드로 로드 (가장 빠르고 효율적!)
    print("✅ 1/2 시도: 스트리밍 모드 (Streaming=True)로 데이터셋을 로드합니다...")
    dataset = load_dataset(DATASET_NAME, split=SPLIT_NAME, streaming=True)
    print("✨ 성공! 스트리밍 데이터셋으로 연결되었습니다. (메모리 효율 최고!)")
    
except Exception as e:
    # 🟡 2차 시도: 스트리밍 실패 시, 일반 Dataset 객체로 로드 (데이터 전체 분석용)
    print(f"\n⚠️ 스트리밍 로드 실패 ({type(e).__name__}): {e}")
    print("   ➡️ 일반 Dataset으로 전환하여 적은 샘플만 로드합니다...")
    try:
        dataset = load_dataset(DATASET_NAME, split=SPLIT_NAME, streaming=False)
        print("✅ 성공! 일반 데이터셋으로 로드되었습니다. (안전 모드 ON)")
    except Exception as e_fallback:
        print(f"❌ 심각한 오류 발생: 모든 로드 시도가 실패했습니다. 라이브러리 또는 네트워크를 확인해주세요. ({e_fallback})")
        exit()


# ------------------------------------------------------------------------------
# 📐 2단계: 샘플링 및 데이터 준비 (핵심 데이터 분석을 위해 작은 조각만!)
# ------------------------------------------------------------------------------

print("\n" + "=" * 80)
print("🔬 2단계: 데이터 샘플을 준비하고 주요 정보를 탐색해 봅시다.")
print("=" * 80)

# 스트리밍 여부에 따라 샘플링 방식을 다르게 적용합니다.
if hasattr(dataset, "take"):
    # .take()가 존재하면 스트리밍 데이터셋 (IterableDataset)
    print("👉 (스트리밍 모드): take() 메서드를 사용하여 상위 샘플만 가져옵니다.")
    sample_iterator = dataset.take(SAMPLE_COUNT)
    
    # Iteration을 위해 리스트로 변환합니다.
    sample_data_list = list(sample_iterator)

else:
    # 일반 데이터셋 (Dataset)
    print("👉 (일반 모드): take() 메서드를 사용하여 상위 샘플만 가져옵니다.")
    # 일반 Dataset의 경우, take() 후 리스트로 변환하는 것이 안전합니다.
    sample_data_list = list(dataset.take(SAMPLE_COUNT))

print(f"\n✅ 최종적으로 {len(sample_data_list)}개의 샘플을 분석 대상으로 선정했습니다!")

# ------------------------------------------------------------------------------
# 💡 3단계: 구조 분석 및 메타데이터 추출 (전문가처럼 데이터 구조 파악하기!)
# ------------------------------------------------------------------------------

print("\n" + "✨" * 10 + " 3단계: 데이터셋의 구조와 내용물을 들여다보기 ✨" + "\n")

# 각 샘플을 반복하며 핵심 메타데이터를 추출합니다.
print("--- 🎤 [샘플 메타데이터 추출 현황] ---")
language_counts = {}
format_counts = {}

for i, sample in enumerate(sample_data_list):
    # 🔑 특징: Dataset 구조는 딕셔너리 형태로 되어 있습니다.
    # 우리가 관심 있는 필드는 'Language'와 'Format'입니다.
    
    language = sample.get('Language', 'N/A')
    file_format = sample.get('Format', 'N/A')
    
    # 📈 정량적 분석 1: 언어 분포 카운트
    language_counts[language] = language_counts.get(language, 0) + 1
    
    # 📈 정량적 분석 2: 파일 포맷 분포 카운트
    format_counts[file_format] = format_counts.get(file_format, 0) + 1
    
    # 오디오 파일 ID만 출력하여 내용물을 확인합니다.
    # 실제로 오디오를 로드하려면 추가 라이브러리가 필요하므로 ID만 출력합니다.
    print(f"[{i+1}/{SAMPLE_COUNT}] ID: {sample.get('ID', 'Unknown')} | Language: {language} | Format: {file_format}")


# ------------------------------------------------------------------------------
# 🧠 4단계: AI 시뮬레이션 - LLM 프롬프트 생성 및 데이터 유형 추론
# ------------------------------------------------------------------------------
# 데이터셋 특성상, '대화 내용'을 분석하는 것이 가장 창의적입니다.
# 우리는 이 메타데이터를 활용하여, 가상의 LLM(거대 언어 모델)이 다음 대화를 추론하는 프롬프트를 설계해봅시다.

print("\n" + "=" * 80)
print("💡 4단계: AI 엔진 작동! (가상의 대화 분석 및 프롬프트 생성)")
print("=" * 80)

# 📊 분석 결과 요약 (정량적 결과 출력)
print("\n[📊 데이터셋 메타 정보 요약 분석]")
print(f"✅ 언어별 분포: {language_counts}")
print(f"✅ 파일 형식 분포: {format_counts}")

# 📝 LLM 프롬프트 설계 시뮬레이션
print("\n[🧠 LLM(대규모 언어 모델) 프롬프트 설계]")
print("----------------------------------------------------")

# 가장 많이 등장한 언어와 포맷을 뽑아 프롬프트에 활용합니다.
most_common_lang = max(language_counts, key=language_counts.get)
most_common_format = max(format_counts, key=format_counts.get)

print("\n👉 목표: AI에게 '이런 특징을 가진 오디오 파일의 대화 스크립트를 분석해줘'라고 요청한다.")

llm_prompt_template = f"""
[시스템 역할 지침]
당신은 전문 '인간-로봇 대화 분석가'입니다. 당신의 임무는 주어진 오디오 파일의 메타데이터(Language: {most_common_lang}, Format: {most_common_format})를 기반으로,
가장 자연스러운 대화 스크립트를 추론하는 것입니다.

[분석 요청]
1. 이 대화가 이루어졌을 법한 '상황(Context)' 3가지.
2. 로봇(AI)이 응답할 때 가장 자연스러운 '다음 문장' 1가지.
3. 대화가 중단된 원인(예: 기술적 문제, 주제 변경 등)을 추론하세요.
"""

print("\n=====================================================")
print("✨ 최종 생성된 LLM 프롬프트 (가장 강력한 분석을 위한 설계):")
print(llm_prompt_template)
print("=====================================================")

print("\n🚀🎉🎉👏✨ 실습 완료! 🎉👏✨🚀")
print("축하합니다! 당신은 데이터셋의 구조를 이해하고, 분석 결과를 바탕으로 AI 엔진에 명령을 내리는,")
print("최고의 AI 연구원이 되었습니다! 👍")
print("==============================================================================")
```

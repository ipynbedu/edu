```
import random
import matplotlib.pyplot as plt
from datasets import load_dataset, get_dataset_config_names
import numpy as np
import pandas as pd

# 🍚🌮🇮🇳 데이터셋 이름: ADVISORYz/Korean_Chinese_Indian_Food
# 🌟 데이터셋 의미: 한국, 중국, 인도 요리에 대한 정보를 모아놓은 데이터셋입니다.
# 📚 목표: 이 데이터셋을 통해 문화적, 요리적 패턴을 AI가 학습하는 과정을 시뮬레이션해보는 실습을 할 거예요!
# 친절한 튜터가 여러분을 도와줄 테니, 코드를 따라와 보세요. 😊

# --- 설정 상수 ---
DATASET_NAME = "ADVISORYz/Korean_Chinese_Indian_Food"
SAMPLE_SPLIT = 'train'
SAMPLE_COUNT = 5 # 초보자 실습이므로, 일단 5개 샘플만 살펴볼게요!

print("="*70)
print(f"👋 안녕하세요! 파이썬으로 맛있는 AI 실습을 시작해 볼 시간이에요.")
print(f"🍔 오늘 우리가 다룰 데이터는 '{DATASET_NAME}' 입니다.")
print("="*70)

# -----------------------------------------------------------------
# 💾 1단계: 데이터셋 로드하기 (스트리밍 모드 테스트)
# -----------------------------------------------------------------

# 🛠️ 주의사항: 데이터 로딩은 가장 까다롭습니다!
# 스트리밍으로 로드하는 것이 가장 빠르고 효율적이지만, 환경에 따라 실패할 수 있어요.
try:
    print("✨ [1/3] 스트리밍 모드로 데이터셋을 로드해 봅니다... (가장 빠름)")
    # streaming=True를 사용하여 메모리를 아끼고 빠르게 데이터를 탐색합니다.
    dataset = load_dataset(DATASET_NAME, split=SAMPLE_SPLIT, streaming=True)
    print("✅ 스트리밍 모드 로드 성공! 메모리 절약에 성공했습니다.")
    LOADED_MODE = "Streaming"

except Exception as e:
    # 🚨 만약 스트리밍 모드에서 문제가 발생하면 (예: 연결 문제, 환경 제약),
    # 작은 크기의 일반 데이터셋으로 대체 로드를 시도합니다.
    print(f"❌ 스트리밍 모드 로드 실패 (오류: {e}). 작은 배치로 대체 로드를 시도합니다.")
    try:
        dataset = load_dataset(DATASET_NAME, split=SAMPLE_SPLIT, streaming=False)
        print("✅ 일반 데이터셋 모드 로드 성공! 작은 테스트 셋을 사용합니다.")
        LOADED_MODE = "Non-Streaming"
    except Exception as e_fail:
        print(f"💔 데이터셋 로드 자체에 심각한 오류가 발생했습니다: {e_fail}")
        exit()

# -----------------------------------------------------------------
# 📋 2단계: 샘플링 및 데이터 준비
# -----------------------------------------------------------------

print("\n" + "="*70)
print("[2/3] 실습에 사용할 샘플 묶음을 준비합니다...")

# 🌠 핵심 패턴: streaming 모드와 일반 모드에 관계없이 .take()를 사용하여 샘플링합니다.
if hasattr(dataset, "take"):
    # .take()가 존재하면 스트리밍 데이터셋(IterableDataset)입니다.
    print(f"💡 샘플러 사용: {SAMPLE_COUNT}개의 샘플만 탐색하겠습니다.")
    # 반복자(iterator) 형태로 변환합니다.
    sampled_dataset_iterator = dataset.take(SAMPLE_COUNT)
else:
    # 일반 데이터셋 (Dataset)일 경우에도 take()가 가능합니다.
    sampled_dataset_iterator = dataset.take(SAMPLE_COUNT)

# 실습을 위해 샘플을 리스트로 변환합니다.
sample_data_list = list(sampled_dataset_iterator)

if not sample_data_list:
    print("⚠️ 경고: 샘플 데이터를 가져올 수 없습니다. 로드 설정을 확인해주세요.")
    exit()


# -----------------------------------------------------------------
# 🧠 3단계: 창의적인 AI 실습 시뮬레이션 (메타데이터 분석)
# -----------------------------------------------------------------

# 우리가 데이터의 내용을 정확히 모르기 때문에, 가장 기본적인 AI 실습인
# '샘플 구조 분석 및 키워드 추출'을 통해 데이터의 패턴을 이해하는 과정을 보여줄게요.
print("\n" + "="*70)
print("✨ [3/3] AI 메타데이터 분석 실습 시작! (Fun Analysis)")
print("   -> 어떤 정보를 추출해야 AI가 똑똑해질까요?")

# 1. 데이터 스키마 확인 (어떤 칼럼이 있는지 확인)
sample_keys = list(sample_data_list[0].keys())
print(f"\n[🔍 데이터 구조 분석]: 이 데이터셋의 주요 키(Column)는 다음과 같습니다: {sample_keys}")
print("   * 분석 목표: 이 키들 중 '요리 카테고리'나 '지역' 같은 것을 찾아봅시다.")

# 2. 반복문을 사용하여 샘플별 내용 출력 및 구조 확인
print("\n[📚 샘플 5개 탐색]: 첫 몇 개의 데이터를 하나씩 꺼내보며 내용을 살펴볼게요.")
for i, sample in enumerate(sample_data_list):
    print(f"  --- Sample {i+1} 내용 ---")
    # 만약 'page_num' 같은 명확한 필드가 있다면, 그것을 보여줍니다.
    if 'page_num' in sample:
        print(f"  * 페이지 번호 (Page ID): {sample['page_num']}")
    else:
        print(f"  * 구조 확인: 키 값들이 {list(sample.keys())} 형태입니다.")
        # 텍스트 내용이 있다면, 그 내용을 간단히 출력합니다. (가정)
        if 'text_content' in sample and sample['text_content']:
            print(f"  * 내용 요약: '{str(sample['text_content'])[:50]}...'")

# 3. 가상의 AI 분석 시뮬레이션: 데이터 분포 시각화 (Mock Visualization)
# 실제 데이터가 없으므로, 분석 과정을 시뮬레이션하여 시각화 방법을 보여줍니다.
print("\n[📈 AI 시각화 실습]: 데이터 분포를 보기 위해 파이썬 라이브러리를 사용하는 법을 알아봅시다.")

# 가상의 3개 요리 카테고리별 분포를 가정합니다.
mock_categories = ['Korean', 'Chinese', 'Indian']
mock_counts = [400, 350, 250]

plt.figure(figsize=(8, 5))
plt.bar(mock_categories, mock_counts, color=['red', 'blue', 'green'])
plt.title('Mock Distribution of Food Categories')
plt.xlabel('Cuisine Type')
plt.ylabel('Number of Items')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()

# -----------------------------------------------------------------
print("\n=====================================================================")
print("🎉🎉🎉 실습 완료! 정말 잘하셨어요! 🎉🎉🎉")
print("이 코드는 데이터 로드, 샘플링, 그리고 분석 시각화의 기본적인 파이썬 AI 워크플로우를 보여줍니다.")
print("다음 시간에는 '어떤 키(key)'를 실제로 추출하고 '텍스트 전처리'를 해볼 거예요!")
print("=====================================================================")
```
